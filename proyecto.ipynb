# B. Análisis de Series de Tiempo

Evolución de `count sheep` (cantidad de ovejas) a lo largo de `[step]`, para cada valor de `initial-number-sheep`.

```r
system('wget "https://raw.githubusercontent.com/juandrodriguezf/proyecto_netlogo/master/wolf_sheep/wolfsheep_evol.csv" -O wolfsheep_evol.csv')
```

```r
data <- read_csv("wolfsheep_evol.csv", skip=6)
head(data)
```

```r
colnames(data) <- c("x", "numOvejas", "tiempo", "cantOvejas")
head(data)
```

```r
data %>% select(numOvejas, tiempo, cantOvejas) -> datas
head(datas)
```

```r
datas %>% group_by(numOvejas, tiempo) %>% summarize(pVivos = mean(cantOvejas)) -> datag
head(datag)
```

```r
datag %>% ggplot(aes(tiempo, pVivos)) + geom_line(aes(group = numOvejas, color = numOvejas))
```

```r
datag %>% ggplot(aes(tiempo, pVivos)) + geom_line(aes(color = numOvejas)) + facet_wrap(~numOvejas)
```
